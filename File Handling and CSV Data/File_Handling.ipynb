### Creating a File :

```python
file = open("students.txt","w") # w = write mode
file.write("reema - 82 marks\n")
file.write("meena - 90 marks\n")
file.write("siya - 72 marks\n")

file.close()
```

### Read a File :

```python
file = open("students.txt","r") # r = read mode
data = file.read()
print(data)
file.close()
```

### with open() :

```python
# with open() = Python automatically file ko properly close kar deta hai.
with open("students.txt","r") as file:
    data = file.read()
    print(data)
```

### Read Multiples Line in a List :

```python
with open("students.txt", "r") as file:
    data = file.readlines()

print(data)
```

### Creating CSV File in Python :

```python
import csv
students = [
    ["reema", 82,87],
    ["meena", 90,81],
    ["siya", 72,78]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Marks", "Attendance"])
    writer.writerows(students)
```

### Read CSV File :

```python
import csv
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
