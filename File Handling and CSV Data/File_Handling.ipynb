### Creating a File :

```python
file = open("students.txt","w") # w = write mode
file.write("reema - 82 marks\n")
file.write("meena - 90 marks\n")
file.write("siya - 72 marks\n")

file.close()
```

### Read a File :

```python
file = open("students.txt","r") # r = read mode
data = file.read()
print(data)
file.close()
```

### with open() :

```python
# with open() = Python automatically file ko properly close kar deta hai.
with open("students.txt","r") as file:
    data = file.read()
    print(data)
```

### Read Multiples Line in a List :

```python
with open("students.txt", "r") as file:
    data = file.readlines()

print(data)
```

### Creating CSV File in Python :

```python
import csv
students = [
    [1,"reema", 82,87],
    [2,"meena", 90,81],
    [3,"siya", 72,78]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Roll No","Name", "Marks", "Attendance"])
    writer.writerows(students)
```

### Read CSV File :

```python
import csv
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# Calculate the average marks of students from the CSV file :
import csv
total_marks = 0
count = 0
with open("students.csv", "r") as file:
    r = csv.DictReader(file)
    for student in r:
        marks = int(student["Marks"])
        total_marks += marks
        count += 1
avg = total_marks / count
print("Average marks of students:", round(avg,2))
```

```python
import csv

students = [
    ["Aman", 82, 88],
    ["Riya", 65, 72],
    ["Karan", 91, 94],
    ["Neha", 48, 58],
    ["Priya", 76, 81]
]
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Marks", "Attendance"])
    writer.writerows(students)

total_marks = 0
count = 0
risk_count = 0

with open("students.csv", "r") as file:

    r= csv.DictReader(file)

    for student in r:

        marks = int(student["Marks"])
        attendance = int(student["Attendance"])

        total_marks += marks
        count += 1

        if attendance < 75 or marks < 60:
            risk_count += 1
            print(student["Name"], "→ At Risk")


average = total_marks / count

print("Student Data Analysis:")
print("Total Students:", count)
print("Average Marks:", round(average, 2))
print("At-Risk Students:", risk_count)
```
