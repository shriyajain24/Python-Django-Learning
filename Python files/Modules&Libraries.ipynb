# Modules :
A module is a Python file (.py) containing reusable Python code.

Creating over own module :

```python
import studentutils

marks = [78,90,80,76,45,71]
avg = studentutils.calculate_avg(marks)
print("Average marks are :",round(avg,2))
```

```python
marks = 80
grade = studentutils.student_grade([marks])
print("Grade:", grade)
```

```python
attendance = 75
marks = 90
status = studentutils.student_attendance(attendance, marks)
print("Student status:", status)
```

```python
# from studentutils import calculate_avg, student_grade, student_attendance
from studentutils import calculate_avg
marks = [78,90,80,76,45,71]
print("Average marks are :",round(calculate_avg(marks),2))
```
