### __init__() Method :


```python
class student:
    def __init__(self,name,marks,attendance):
        self.name = name
        self.marks = marks
        self.attendance = attendance
```

```python
student1 = student("reema",90,86)
```

```python
print(student1.name)
print(student1.marks)
print(student1.attendance)
```

```python
def student_marks(self):
    if self.marks >= 90:
        return "A"
    elif self.marks >= 80:
        return "B"
    elif self.marks >= 70:
        return "C"
    elif self.marks >= 60:
        return "Average"
    else:
        return "Fail"
```

```python
student1.student_marks = student_marks.__get__(student1)
```

```python
print(student1.student_marks())
```

```python
print(student_marks(student1))
```
