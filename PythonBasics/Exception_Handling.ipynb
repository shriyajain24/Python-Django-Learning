### Basic try-except :

```python
try:
    marks1 = int(input("Enter your marks: "))
    print("Marks 1 is :",marks1)

except ValueError:
    print("Please enter valid input")

try:
    marks2 = int(input("Enter your marks: "))
    print("Marks 2 is :",marks2)
except ValueError:
    print("Please enter valid input")
```

### Multiple Exceptions :

```python
try:
    number1 = int(input("Enter your number: "))
    number2 = int(input("Enter your number: "))
    result = number1 / number2
    print("Result is :",result)

except ValueError:
    print("Please enter valid input")
except ZeroDivisionError:
    print("Number can't be divided by zero")
```

### finally :

```python
try:
    marks = int(input("Enter your marks: "))
    print("Marks is :",marks)

except ValueError:
    print("Please enter valid input")

finally:
    print("Student Marks")
```

### raise :

```python
# raise = generate our own error

def student_marks(marks):
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100")
    return marks

try:
    marks = student_marks(int(input("Enter your marks: ")))
    print("Marks is :",marks)
except ValueError as error:
    print("Invalid input:", error)
    
```

### Mini Project :

```python
students = [
    {"name": "Aman", "marks": 82, "attendance": 88},
    {"name": "Riya", "marks": "abc", "attendance": 72},
    {"name": "Karan", "marks": 91, "attendance": 94},
    {"name": "Neha", "marks": 48, "attendance": 58},
    {"name": "Priya", "marks": 76, "attendance": 81}
]


def student_marks(marks):

    marks = float(marks)

    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    return marks


def student_info(students):

    for student in students:

        try:
            marks = student_marks(student["marks"])

            print("-" * 30)
            print("Name:", student["name"])
            print("Marks:", marks)
            print("Attendance:", student["attendance"], "%")

            if marks >= 75:
                print("Category: Good")
            elif marks >= 60:
                print("Category: Average")
            else:
                print("Category: Needs Improvement")

        except ValueError as error:

            print("-" * 30)
            print("Name:", student["name"])
            print("Data Error:", error)


student_info(students)
```
