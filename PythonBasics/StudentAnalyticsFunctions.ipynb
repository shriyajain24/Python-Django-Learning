```python
students = [
    {"name": "Aman", "marks": 82, "attendance": 88},
    {"name": "Riya", "marks": 65, "attendance": 72},
    {"name": "Karan", "marks": 91, "attendance": 94},
    {"name": "Neha", "marks": 48, "attendance": 58},
    {"name": "Priya", "marks": 76, "attendance": 81}
]

def calculate_average_marks(students):
    total = 0
    for student in students:
        total += student["marks"]
        average = total / len(students)
    return average

def highest_score(students):
    highest = students[0]
    for student in students:
        if student["marks"] > highest["marks"]:
            highest = student
    return highest

def lowest_score(students):
    lowest = students[0]
    for student in students:
        if student["marks"] < lowest["marks"]:
            lowest = student
    return lowest

def attendance_count(students):
    count = 0
    for student in students:
        if student["attendance"] >= 75:
            count = count + 1
    return count

def student_info(students):
    calculate_average = calculate_average_marks(students)
    highest = highest_score(students)   
    lowest = lowest_score(students)
    attendance = attendance_count(students)

    print("----------------Student Information-----------------")
    print(f"Total Students: {len(students)}")   
    print(f"Average Marks: {calculate_average}")
    print(f"Highest Score: {highest['name']} with {highest['marks']} marks")
    print(f"Lowest Score: {lowest['name']} with {lowest['marks']} marks")
    print(f"Number of students with attendance >= 75% : {attendance}")   

student_info(students)

```
