```python
students = [
    {"name": "Siya", "attendance": 88, "marks": 82, "assignment": 90},
    {"name": "Sonal", "attendance": 91, "marks": 90, "assignment": 94},
    {"name": "Neha", "attendance": 58, "marks": 48, "assignment": 55},
    {"name": "Divya", "attendance": 79, "marks": 76, "assignment": 80}
]


def final_score(marks, assignment):
    return (marks * 0.70) + (assignment * 0.30)


def get_category(score):
    if score >= 90:
        return "Excellent"
    elif score >= 75:
        return "Good"
    elif score >= 60:
        return "Average"
    else:
        return "Needs Improvement"


def check_risk(attendance, score):
    if attendance > 75 or score > 60:
        return "Safe"
    else:
        return "Needs Attention"


for student in students:

    score = final_score(
        student["marks"],
        student["assignment"]
    )

    category = get_category(score)
    performance = check_risk(student["attendance"], score)

    print("--------------------------------")
    print("Name:", student["name"])
    print("Attendance:", student["attendance"], "%")
    print("Final Score:", round(score, 2))
    print("Category:", category)
    print("Status:", performance)
```
