PRACTICAL 1 — Student Information

```python
name = input("Enter your name :")
age = int(input("Enter your age :"))
attendance = float(input("Enter your attendance :"))
marks = float(input("Enter your marks :"))
assignment = float(input("Enter your assignment marks :"))  

print("---------------Student Information-----------------")
print("Name is :" , name)
print("Age is :",age)
print("Attendance is :",attendance,"%")
print("Marks is :",marks)
print("Assignment marks is :",assignment)
```

PRACTICAL 2 — Attendance Status

```python
if attendance >= 75:
    print("Attendance is good")
else:
    print("Attendance is Low")
```

PRACTICAL 3 — Student Performance Analyzer

```python
name = input("Enter your name :")
age = int(input("Enter your age :"))
attendance = float(input("Enter your attendance :"))
marks = float(input("Enter your marks :"))
assignment = float(input("Enter your assignment marks :"))  

score = (attendance + assignment + marks) / 3

if score >= 80:
    level = "Excellent"
elif score >= 60:
    level = "Good"
elif score >= 40:
    level = "Average"
else:
    level = "Needs Improvement"

print("---------------Student Information-----------------")
print("Student Name is :", name)
print("Age of Student is :", age)
print("Student Attendance is :", attendance, "%")
print("Student Marks is :", marks)
print("Student Assignment Marks is :", assignment)
print("Student Overall Score is :", round(score,2))
print("Student Performance Level is :", level)

if attendance >= 75:
    print("Attendance is good")
else:
    print("Attendance is Low")
```
